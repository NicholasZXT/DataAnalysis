### 损失函数

sklearn里GBDT支持的损失函数如下.

#### 回归问题
`GradientBoostingRegressor`
+ squared_error，默认值
+ absolute_error
+ quantile
+ huber

`HistGradientBoostingRegressor`
+ squared_error
+ absolute_error
+ quantile
+ gamma
+ poisson

各损失函数详情如下：

+ **squared_error——平方误差**

$$
L = \sum{(y_i - H(x_i))^2}
$$

+ **absolute_error——绝对误差**

$$
L = \sum{|y_i - H(x_i)|}
$$


+ **quantile**

$$
L = \sum{l(y_i,H(x_i))}
$$

其中
$$
l = \begin{split} 
\begin{cases}
    \alpha (y_i - H(x_i)), & y_i - H(x_i) > 0 \\
    0,    & y_i - H(x_i) = 0 \\
    (1-\alpha) (y_i - H(x_i)), & y_i - H(x_i) < 0
\end{cases}\end{split}, \space \space \alpha \in (0, 1)
$$


+ **huber——合页误差**

$$
L = \sum{l(y_i,H(x_i))}
$$

其中
$$
l = \begin{split} 
\begin{cases}\frac{1}{2}(y_i - H(x_i))^2, & |y_i - H(x_i)|\leq\alpha \\
\alpha(|y_i - H(x_i)|-\frac{\alpha}{2}),& |y_i - H(x_i)|>\alpha \end{cases}\end{split}, \space \space \alpha \in (0, 1)
$$

其中$\alpha$是需要我们自己设置的超参数，由参数`alpha`控制。在huber损失中，alpha是阈值，在quantile损失中，alpha用于辅助计算损失函数的输出结果，默认为0.9

#### 分类问题
`GradientBoostingClassifier`
+ log_loss
+ exponential

`HistGradientBoostingClassifier`
+ log_loss, 只支持这一种


对任意样本$i$而言，$y_i$为真实标签，$\hat{y_i}$为预测标签，$H(x_i)$为集成算法输出结果，$p(x_i)$为基于$H(x_i)$和sigmoid/softmax函数计算的概率值。
那么各损失函数详情如下：

+ **log_loss——对数几率**，也叫交叉熵损失

**二分类交叉熵损失**
$$
L = -\left( y\log p(x) + (1 - y)\log(1 - p(x)) \right)
$$
注意，log当中输入的一定是概率值。对于逻辑回归来说，概率就是算法的输出，因此我们可以认为逻辑回归中$p = H(x)$，但对于GBDT来说，$p(x_i) = Sigmoid(H(x_i))$，这一点一定要注意。


**多分类交叉熵损失**，总共有K个类别

$$
L = -\sum_{k=1}^Ky^*_k\log(P^k(x))
$$

其中，$P^k(x)$是概率值，对于多分类GBDT来说，$p^k(x) = Softmax(H^k(x))$，$y^*$是由真实标签转化后的向量。

+ **exponential——指数损失**

**二分类指数损失**
$$
L = e^{-yH(x)}
$$

**多分类指数损失**，总共有K个类别
$$
\begin{aligned}
L &=exp \left( -\frac{1}{K}\boldsymbol{y^* · H^*(x)} \right) = exp \left( -\frac{1}{K}(y^1H^1(x)+y^2H^2(x) \ + \  ... + y^kH^k(x)) \right)
\end{aligned}
$$

需要注意，指数损失中的$y^*$与交叉熵损失中的$y^*$不是同样的向量。

## Stacking - KEY

有关Stacking的介绍，可以参考如下的几篇博客：
1. [模型融合方法概述](https://zhuanlan.zhihu.com/p/25836678)
2. [Kaggle机器学习之模型融合（stacking）心得](https://zhuanlan.zhihu.com/p/26890738)
3. [Introduction to Ensembling/Stacking in Python](https://www.kaggle.com/arthurtok/introduction-to-ensembling-stacking-in-python/notebook#Second-Level-Predictions-from-the-First-level-Output)

实现类详见`Stacking.py`文件。

sklearn提供的API如下：
```python
# 回归问题
StackingRegressor(
    estimators, 
    final_estimator=None, 
    *, 
    cv=None, 
    n_jobs=None, 
    passthrough=False, 
    verbose=0
)

# 分类问题
StackingClassifier(
    estimators, 
    final_estimator=None, 
    *, 
    cv=None, 
    stack_method='auto', 
    n_jobs=None, 
    passthrough=False, 
    verbose=0
)
```

**参数**：
+ `estimators`: list of `(str, estimator)`
+ `final_estimator`: 元估计器
+ `cv`: 设置用于**训练final_estimator**时交叉验证的策略，这个参数要注意一下，可选值如下：
    + `None`，使用默认的5折交叉验证
    + 整数，指定交叉验证折数
    + 可以执行cross-validation的generator
    + 分割数据集的可迭代对象
    + `prefit`: <font color='red'>这个取值要特别注意，此时表示所有的一级估计器都是训练好的，元估计器不会使用交叉验证策略，而是使用一级估计器在所有数据上的输出</font>，此时会有数据泄露的情况，非常容易出现过拟合的情况。
+ `stack_method`: 融合策略（<font color='red'>只针对`StackingClassifier`有效</font>），可选值为`{'auto', 'predict_proba', 'decision_function', 'predict'}`
    + `auto`下会对每个估计器依次尝试`predict_proba`, `decision_function`, `predict` 方法
    + 其他选项则是自己指定使用的估计器方法
+ `passthrough`: 默认`False`，表示只有一级估计器的输出会被输入到元估计器中，`True`表示则会和原始数据一起输入元估计器中。

**常用方法**：

- `transform(X)`: 输出所有**一级`estimators`的X结果**（不是预测结果），形状是`(n_samples, n_estimators)`
- `fit(X, y, **fit_params)`: 训练一级估计器和二级估计器
- `fit_transform(X, y, **fit_params)`: 等价于 `.fit(X, y).transform(X)` 的链式调用
- `predict(X, **predict_params)`: 预测 `X` 结果
- `score(X, y, sample_weight=None)`
- `get_params(deep=True)`: 以dict形式返回参数，`deep=True`时也会一并返回所有`estimators`里的参数。
- `set_params`: 设置参数，通过`get_param`返回的key，也可以**传入一级估计器的参数**。

> 顺带一提，`GridSearchCV.fit()` 方法里，也是调用 `estimator.set_params()` 方法设置参数的，也就是说可以把 `StackingClassifier` 传递给 `GridSearchCV`，并且进行 一级estimator 和 final_estimator 的超参数搜索。

### 说明

通过查看源码，Sklearn里实现的Stacking（`StackingRegressor`和`StackingClassifier`）有如下几个需要注意的点：

（1）传入的一级估计器`estimators`，都被认为是已经配置好了 **<font color='red'>超参数</font>** 的估计器，后续的 fit 中，只会对普通参数进行训练。

（2）如果`cv='prefit'`:
+ `fit`方法里不会对一级学习器进行训练（普通参数），也不会使用`cross_val_predict`获取一级学习器的OOB预测，而是直接使用每个一级学习器在整个数据集上的预测结果作为元学习器的训练数据。
+ `predict`方法里没有啥特殊处理

（3）如果`cv`取值不是`prefit`：
+ `fit`方法里首先<font color='red'>会**对每个一级学习器**使用传入的**整个数据集**训练一次模型参数（**不是超参数**）</font>。
+ 然后使用交叉预测（`cross_val_predict`）的方式得到每个一级学习器的OOB预测，作为元学习器的训练数据，但是`cross_val_predict`过程中**每折数据训练普通参数的模型不会保存**。
+ 这一意味着，<font color='red'>传入的一级学习器所使用的最优超参数需要实现我们自己实现执行一次GridSearch+CV的搜索</font>，但是这个过程会出现数据泄露的情况，导致后续使用Stacking集成时极易出现过拟合的情况。
+ `predict`方法里没有特殊处理

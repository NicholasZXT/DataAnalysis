# Sklearn

`sklearn.model_selection`中也提供了超参数搜索的API，主要有如下4个：
+ `GridSearchCV`: 全网格搜索
+ `RandomizedSearchCV`: 随机网格搜索
+ `HalvingGridSearchCV`: 对半网格搜索
+ `HalvingRandomSearchCV`: 对半网格随机搜索

以下是使用的数据示例。

```python
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
import pandas as pd
```

```python
X, y = make_moons(n_samples=280, noise=0.1, random_state=42)
```

```python
fig = plt.figure(figsize=(6, 4))
ax = fig.add_subplot(111)
ax.set_title("faked moon data")
ax.scatter(X[:, 0], X[:,1], c=y)
```

## GridSearchCV

官方API文档 [GridSearchCV](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html)。

常用参数如下：
+ `estimator`: 待优化的模型，必须是scikit-learn的模型
+ `param_grid`: 参数空间，是一个字典，key是参数名，value是参数的取值范围
+ `scoring`: 评估函数
+ `cv`: 交叉验证的参数，默认为5；也可以是一个cross-validation generator
+ `refit`: 是否在搜索结束后重新使用全部数据训练模型，默认为True
+ `return_train_score`: 是否返回训练集上的评估结果，默认为False
+ `n_jobs`: 并行搜索的进程数，默认为1
+ `verbose`:输出信息的等级，默认为0

常用属性如下：
+ `best_estimator_`: 最优的模型timator`: 最优的模型
+ `best_score_`: 最优的评估结果
+ `best_params_`: 最优的参数
+ `best_index_`: 最优的索引
+ `scorer_`: 评估函数
+ `cv_results_`: 交叉验证的结果

常用方法如下：
+ `fit`: 执行搜索，返回一个`GridSearchCV`对象
+ `predict`
+ `score`
+ `predict_proba`
+ `predict_log_proba`
+ `decision_function`

GridSearchCV是最基本的网格搜索方法，相当于穷举搜索。

sklearn也提供了一些网格搜索优化方法，主要包括两类，其一是**调整搜索空间**，其二是**调整每次训练的数据**。

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc = RandomForestClassifier()
param_grid = {
    'criterion': ['gini', 'entropy'],
    'n_estimators': [10, 20, 30, 40, 50],
    'max_depth': [3, 4, 5],
    'min_impurity_decrease': [0.5, 1.0, 1.5]
}
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
# gs = GridSearchCV(estimator=rfc, param_grid=param_grid, cv=skf, scoring='accuracy', refit=True, return_train_score=True, n_jobs=-1)
# 不设置 n_job = -1 时，默认使用单核，会慢不少
gs = GridSearchCV(estimator=rfc, param_grid=param_grid, cv=skf, scoring='accuracy', refit=True, return_train_score=True)
```

```python
%%timeit -n 1 -r 1  # 执行1次就行了，不需要重复执行
gs.fit(X, y)
```

```python
gs.best_score_
```

```python
gs.best_params_
```

```python
gs.best_estimator_
```

```python
gs.best_index_
```

```python
cv_res = pd.DataFrame(gs.cv_results_)
print("cv_res.shape: ", cv_res.shape)
cv_res
```

## RandomizedSearchCV

官方API文档 [RandomizedSearchCV](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.RandomizedSearchCV.html)。

RandomizedSearchCV 是通过调整搜索参数空间来提高速度的，下面的 `n_iter` 代表的就是在参数空间中不放回抽样的次数。

常用参数如下：
+ `estimator`: 待优化的模型，必须是scikit-learn的模型
+ `param_distributions`: 参数空间
+ `n_iter`: 搜索的次数，默认为10
+ `scoring`: 评估函数
+ `cv`: 交叉验证的参数，默认为5；也可以是一个cross-validation generator
+ `refit`: 是否在搜索结束后重新使用全部数据训练模型，默认为True
+ `return_train_score`: 是否返回训练集上的评估结果，默认为False

常用属性如下：
+ `best_estimator_`: 最优的模型
+ `best_score_`: 最优的评估结果
+ `best_params_`: 最优的参数
+ `best_index_`: 最优的索引
+ `scorer_`: 评估函数
+ `cv_results_`: 交叉验证的结果

常用方法如下：
+ `fit`: 执行搜索，返回一个`RandomizedSearchCV`对象
+ `predict`
+ `score`
+ `predict_proba`


```python
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc = RandomForestClassifier()
param_dis = {
    'criterion': ['gini', 'entropy'],
    'n_estimators': [10, 20, 30, 40, 50],
    'max_depth': [3, 4, 5],
    'min_impurity_decrease': [0.5, 1.0, 1.5]
}
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
rs = RandomizedSearchCV(estimator=rfc, param_distributions=param_dis, cv=skf, scoring='accuracy', refit=True, return_train_score=True, n_jobs=-1)
```

```python
%%timeit -n 1 -r 1
rs.fit(X, y)
```

```python
rs.best_score_
```

## HalvingGridSearchCV

官方API文档 [HalvingGridSearchCV](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.HalvingGridSearchCV.html)。

HalvingGridSearchCV 是对 GridSearchCV 的改进，它通过调整每次的搜索数据来提高速度，大致流程如下：

1. 首先从全数据集中 **无放回随机抽样** 出一个很小的子集 $d_0$，并在 $d_0$ 上验证全部参数组合的性能，根据 $d_0$ 上的验证结果，淘汰评分排在后1/2的那一半参数组合。
2. 然后，从全数据集中再**无放回抽样**出一个比 $d_0$ 大一倍的子集 $d_1$，并在 $d_1$ 上验证剩下的那一半参数组合的性能。根据 $d_1$ 上的验证结果，淘汰评分排在后1/2的参数组合。
3. 重复2，直到备选参数组合只剩下一组，或剩余可用的数据不足。

常用参数如下：
+ `estimator`: 待优化的模型，必须是scikit-learn的模型
+ `param_grid`: 参数空间，是一个字典，key是参数名，value是参数的取值范围
+ `scoring`: 评估函数
+ `cv`: 交叉验证的参数，默认为5；也可以是一个cross-validation generator
+ `refit`: 是否在搜索结束后重新使用全部数据训练模型，默认为True
+ `return_train_score`: 是否返回训练集上的评估结果，默认为False
+ `factor`: 对半因子，指定每轮迭代后留下的参数组合的比例，默认为3，表示每轮迭代后留下的参数组合的比例为1/3
+ `resource`: 设置每轮迭代中增加的验证资源的类型，输入为字符串。默认是样本量，输入为"n_samples"
+ `max_resources`
+ `min_resources`
+ `aggressive_elimination`

> 截止到 1.6.1 版本，HalvingGridSearchCV 还处于实验状态，需要使用 `sklearn.experimental.enable_halving_search_cv` 启用。

```python
from sklearn.experimental import enable_halving_search_cv
from sklearn.model_selection import StratifiedKFold, HalvingGridSearchCV
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc = RandomForestClassifier()
param_grid = {
    'criterion': ['gini', 'entropy'],
    'n_estimators': [10, 20, 30, 40, 50],
    'max_depth': [3, 4, 5],
    'min_impurity_decrease': [0.5, 1.0, 1.5]
}
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
hgs = HalvingGridSearchCV(estimator=rfc, param_grid=param_grid, cv=skf, scoring='accuracy', refit=True, return_train_score=True, n_jobs=-1)
```

```python
%%timeit -n 1 -r 1
hgs.fit(X, y)
```

```python
hgs.best_score_
```

```python
hgs.best_params_
```

# Hyperopt

hyperopt的使用步骤如下：
1. 定义一个待优化的目标函数，要求是 minimize 类型
2. 定义参数搜索空间
3. 指定存储搜索参数存储的数据库——这一步是可选的
4. 指定搜索算法

## 简单示例

```python
from hyperopt import hp, fmin, tpe, Trials, STATUS_OK
from hyperopt.early_stop import no_progress_loss
```

（1）定义待优化的目标函数

```python
def obj_fun_simple(space):
    """space是参数空间，对应 fmin() 函数中 space= 参数的传入值，可以是dict，也可以是单个值"""
    # print("x_space: ", space)
    x = space
    return (x - 5)**2
```

（2）定义参数空间

```python
x_space = hp.uniform('x', 0, 10)
```

（3）执行优化过程，这里跳过了存储数据库

```python
best = fmin(fn=obj_fun_simple, space=x_space, algo=tpe.suggest, max_evals=50)
```

```python
print(best.__class__)
```

```python
best
```

## 复杂示例

定义目标函数时，除了像上面返回一个目标函数值，也可以返回一个dict，用来存放更加丰富的信息。

不过此时需要在返回的dict中设置如下两个特殊的key：
+ `status`，通常是`hyperopt.STATUS_{STRING}`，用于表示此次状态，只有`ok`的记录会被输出
+ `loss`, 浮点数，存放目标函数的值

其他一些可选的key有：
+ `loss_vairance`
+ `true_loss`
+ `true_loss_variance`
+ `attachments`

此外的一个要求就是返回的dict能够被JSON序列化。

为了拿到返回的dict里的自定义信息，需要使用`Trials`对象，它有如下几个封装的属性/方法：
+ `trials`: list of dict, 存放每次迭代时使用的搜索信息
+ `results`: list of dict, 每个dict就是一次迭代中目标函数的返回值
+ `losses()`: list of loss，其中每个loss对应于status=ok的记录
+ `statuses()`: list of status string

```python
def obj_fun(space):
    # print("x_space: ", space)
    x = space
    # 返回一个字典，除了两个必须的key，还有一个要求就是需要能JSON序列化
    return {'loss': (x - 5)**2, 'status': STATUS_OK, 'x_value': x}
```

```python
x_space = hp.uniform('x', 0, 10)
```

```python
trials = Trials()
```

```python
best = fmin(fn=obj_fun, space=x_space, algo=tpe.suggest, max_evals=10, trials=trials)
```

## 参数空间定义

hyperopt里参数空间的定义是通过 `hyperopt.hp` 模块提供的函数实现的，常用函数如下：

+ `hp.randint(label, upper)`    
从 $[0, upper)$ 区间里随机采样。


+ `hp.uniform(label, low, high)`    
从 $[low, high]$ 区间里均匀采样。

+ `hp.quniform(label, low, high, q)`    
返回 round(uniform(low, high) / q) * q 的离散采样值。

+ `hp.loguniform(label, low, high)`    
从 $[exp(low), exp(high)]$ 区间里均匀采样，相当于 exp(uniform(low, high))。

+ `hp.qloguniform(label, low, high, q)`   
相当于 round(exp(uniform(low, high)) / q) * q。

+ `hp.normal(label, mu, sigma)`    
从正态分布 $N(\mu, \sigma)$ 里采样。

+ `hp.qnormal(label, mu, sigma, q)`   
相当于 round(normal(mu, sigma) / q) * q。

+ `hp.lognormal(label, mu, sigma)`   
相当于 exp(normal(mu, sigma))。

+ `hp.qlognormal(label, mu, sigma, q)`   
相当于 round(exp(normal(mu, sigma)) / q) * q 。

+ `hp.choice(label, options)`   
离散采样，`options`是一个list, 随机返回其中的一个项目，对于字符串类型的参数通常使用这个。


上述函数返回只是一个变量，并不是立马完成采样过程，它们只是定义了如何从参数空间中采样，实际采样过程是由 `hyperopt.pyll.stochastic` 模块实现的。

# Optuna

Github地址 [Optuna: A hyperparameter optimization framework](https://github.com/optuna/optuna).

相关教程文档：
+ [Optuna英文文档](https://optuna.readthedocs.io/en/stable/)
+ [Optuna中文文档](https://optuna.readthedocs.io/zh-cn/latest/)，这个的版本更新会有些落后，但是也有参考意义

Optuna使用了如下两个概念：
+ Study: optimization based on an objective function
+ Trial: a single execution of objective function

> The goal of a study is to find out the optimal set of hyperparameter values (e.g., classifier and svm_c) through multiple trials

Optuna的使用和Hyperopt不一样，它的参数空间定义是写在目标函数里面的，而不是单独定义。
一般分为3步：
1. 使用 `optuna.study.create_study` 创建一个 `Study` 对象
2. 定义目标函数，并在目标函数里通过`optuna.trial`提供的一系列方法设置参数空间
3. 使用 `optuna.optimize` 运行优化过程，获取结果

Optuna被设计成使用某个数据库作为后端来存储搜索过程，并实现并行搜索的。

```python
from optuna import create_study
from optuna.trial import Trial
```

（1）定义 Study 对象

```python
# 设置日志输出，可选项
# optuna.logging.get_logger("optuna").addHandler(logging.StreamHandler(sys.stdout))

study_name = "example-study"  # Unique identifier of the study.
# 设置存储的数据库后端
# storage_name = "sqlite:///{}.db".format(study_name)
# 如果设置为 None 或者不传入，就会存放在内存中，不进行持久化
storage_name = None

study = create_study(study_name=study_name, storage=storage_name)
```

```python
study
```

（2）定义目标函数

```python
def obj_fun_simple(trial: Trial):
    # 目标函数的参数是 trial 对象，里面封装了一系列生产参数空间的方法
    x = trial.suggest_float("x", -10, 10)
    # print(x)
    # 使用 Trial 对象记录自定义的信息
    trial.set_user_attr('x_value', x)
    return (x - 2) ** 2
```

（3）运行优化过程

```python
study.optimize(obj_fun_simple, n_trials=5)
```
